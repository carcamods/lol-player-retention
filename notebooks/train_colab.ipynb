# Train the neural nets on Colab

A thin launcher (ADR 0004): it installs this package and runs the same commands as on a laptop. All logic lives in `src/lolret/`. CPU is enough at the size of run `v1` (a few minutes), so this is optional.

Before running:

1. Put `examples.parquet` and `sequences.parquet` from `data/processed/v1/` in a Google Drive folder `lolret/data/processed/v1/`. They hold PUUIDs: keep the folder private.
2. While the repo is private, add a GitHub token with read access as the Colab Secret `GH_TOKEN` (key icon in the left sidebar). Never paste it into a cell.

```
import os

from google.colab import drive, userdata

drive.mount("/content/drive")
os.environ["GH_TOKEN"] = userdata.get("GH_TOKEN")
os.environ["DATA_DIR"] = "/content/drive/MyDrive/lolret/data"
```

```
%%bash
git clone -q https://x-access-token:${GH_TOKEN}@github.com/carcamods/lol-player-retention.git
pip install -q -e lol-player-retention  # Colab already has torch
```

```
%%bash
cd lol-player-retention
lolret train -c configs/train.yaml
lolret train-nn -c configs/train_nn.yaml
cat "$DATA_DIR/models/v1/model_card.md"
```
